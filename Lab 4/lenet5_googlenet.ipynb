```python
import tensorflow as tf
import numpy as np
import time
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import (
    Conv2D, MaxPooling2D, AveragePooling2D,
    Dense, Dropout, Flatten,
    GlobalAveragePooling2D,
    Input, concatenate
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.size": 15,
    "axes.titlesize": 15,
    "axes.labelsize": 15,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "legend.fontsize": 15
})

import warnings

warnings.filterwarnings(
    "ignore",
    message="Do not pass an `input_shape`/`input_dim` argument",
    category=UserWarning
)
```

LeNet 5

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)

print(x_train.shape)
print(x_test.shape)
```

```python
def create_lenet():

    model = Sequential([

        Conv2D(
            6,
            kernel_size=(5,5),
            activation='tanh',
            input_shape=(32,32,3)
        ),

        AveragePooling2D(
            pool_size=(2,2)
        ),

        Conv2D(
            16,
            kernel_size=(5,5),
            activation='tanh'
        ),

        AveragePooling2D(
            pool_size=(2,2)
        ),

        Flatten(),

        Dense(
            120,
            activation='tanh'
        ),

        Dense(
            84,
            activation='tanh'
        ),

        Dense(
            10,
            activation='softmax'
        )
    ])

    return model
```

```python
lenet = create_lenet()

lenet.compile(
    optimizer=Adam(0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

lenet.summary()
```

```python
start=time.time()

lenet_history = lenet.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

lenet_time=time.time()-start
```

```python
lenet_acc = lenet.evaluate(
    x_test,
    y_test
)[1]


print("LeNet Accuracy:",lenet_acc)
print("Training Time:",lenet_time)
print("Parameters:",lenet.count_params())
```

GoogleNet

```python
def inception_block(x):

    conv1 = Conv2D(
        32,
        (1,1),
        activation='relu',
        padding='same'
    )(x)


    conv3 = Conv2D(
        32,
        (3,3),
        activation='relu',
        padding='same'
    )(x)


    conv5 = Conv2D(
        16,
        (5,5),
        activation='relu',
        padding='same'
    )(x)


    pool = MaxPooling2D(
        (3,3),
        strides=(1,1),
        padding='same'
    )(x)


    pool = Conv2D(
        16,
        (1,1),
        activation='relu',
        padding='same'
    )(pool)


    return concatenate(
        [
            conv1,
            conv3,
            conv5,
            pool
        ]
    )
```

```python
def create_googlenet():

    inputs=Input(
        shape=(32,32,3)
    )


    x=Conv2D(
        64,
        (3,3),
        padding='same',
        activation='relu'
    )(inputs)


    x=inception_block(x)

    x=inception_block(x)


    x=GlobalAveragePooling2D()(x)


    x=Dense(
        256,
        activation='relu'
    )(x)


    x=Dropout(0.4)(x)


    outputs=Dense(
        10,
        activation='softmax'
    )(x)


    return Model(
        inputs,
        outputs
    )
```

```python
googlenet=create_googlenet()


googlenet.compile(
    optimizer=Adam(0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

googlenet.summary()
```

```python
start=time.time()


google_history=googlenet.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)


google_time=time.time()-start
```

```python
google_acc=googlenet.evaluate(
    x_test,
    y_test
)[1]


print("GoogleNet Accuracy:",google_acc)
print("Training Time:",google_time)
print("Parameters:",googlenet.count_params())
```
