```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Step Activation function
def step (z):
  if z>=0:
    return 1

  else:
    return 0
```

```python
#Decision boundary for XOR
def decision_boundary(x,y,w,b,title):

    label0=True
    label1=True
    plt.figure(figsize=(5,5))
    for i in range(len(x)):
        if y[i]==0:
          if label0:
            plt.scatter(x[i][0],x[i][1],color='red',label="Output=0")
            label0=False
          else:
            plt.scatter(x[i][0],x[i][1],color='red')
        else:
          if label1:
            plt.scatter(x[i][0],x[i][1],color='blue',label="Output=1")
            label1=False
          else:
            plt.scatter(x[i][0],x[i][1],color='blue')
    x_cord=np.linspace(-0.5,1.5,100)

    if w[1]!=0:
        boundary=-(w[0]*x_cord+b)/w[1]
        plt.plot(x_cord,boundary,color='black',label="Decision Boundary")
    elif w[0]!=0:
        plt.axvline(-b/w[0],color='black',label='Decision Boundary')
    plt.title(title)
    plt.xlabel("Input 1")
    plt.ylabel("Input 2")
    plt.legend()
    plt.savefig(f"{title}.eps",dpi=600)
    plt.show()
```

```python
def perceptron(x,y,gate):
    w=np.zeros(2)
    b=0
    lr=1
    update=1
    converged=False
    for epoch in range(15):
        error=False
        for i in range(len(x)):
            z=np.dot(w,x[i])+b
            pred=step(z)
            err=y[i]-pred
            if err!=0:
                w=w+(lr*err*x[i])
                b=b+(lr*err)

                print(f"\n\nUpdate {update} (Epoch {epoch+1})")
                print("Weights =",w)
                print("Bias =",b)

                update+=1
                error=True
        if not error:
          converged=True
          print(f"Converged at {epoch+1} epoch")
          break
    if error:
        print("Not converged.")

    print("\nFinal Weights:",w)
    print("Final Bias:",b)

    decision_boundary(x,y,w,b,gate+f" (Update= {update-1} and Epoch= {epoch+1})")

    print("\nPredictions")
    for i in x:
        print(i," | ",step(np.dot(w,i)+b))
```

```python
#XOR Gate
x=np.array([[0,0],[0,1],[1,0],[1,1]])
y=np.array([0,1,1,0])

print("XOR Gate")
perceptron(x,y,"XOR")
```
